## What is Graph RAG?

**Graph RAG** extends standard RAG by using knowledge graphs for retrieval. Instead of only using vector similarity, Graph RAG:
1. Retrieves relevant subgraphs from knowledge graph
2. Uses graph structure to find related information
3. Combines graph retrieval with vector retrieval (hybrid)
4. Provides structured context to LLM

### Why Graph RAG?

**Standard RAG Limitations:**
- Vector search finds similar documents
- May miss related information not in query
- Doesn't leverage relationship structure
- Limited to direct similarity

**Graph RAG Benefits:**
- ✅ Finds related information through graph traversal
- ✅ Leverages relationship structure
- ✅ Better for relationship queries
- ✅ Can find multi-hop connections
- ✅ More comprehensive context

### Key Concepts

**Subgraph Retrieval**: Extract relevant portions of graph
- Start from query-matched nodes
- Traverse relationships to find related nodes
- Include nodes within N hops
- Return connected subgraph

**Graph Traversal**: Following relationships
- 1-hop: Direct neighbors
- 2-hop: Neighbors of neighbors
- Multi-hop: Deeper relationships
- Path finding: Shortest paths between entities

**Hybrid Retrieval**: Combining vector + graph
- Vector search: Find semantically similar documents
- Graph search: Find related entities through relationships
- Combine results: Merge both retrieval methods
- Rank and deduplicate: Best of both worlds

**Graph Formatting**: Convert graph to text for LLM
- Extract node properties
- Include relationship information
- Format as natural language
- Structure for LLM understanding

### When to Use Graph RAG

✅ **Good for:**
- Knowledge bases with rich relationships
- Queries about entity relationships
- When you need multi-hop reasoning
- Complex queries requiring graph traversal
- When standard RAG misses related information
- Domain-specific knowledge graphs

❌ **Not ideal for:**
- Simple factual queries (standard RAG sufficient)
- When relationships aren't important
- Very large graphs (traversal can be slow)
- When you don't have a knowledge graph
- Real-time applications with strict latency requirements


## Theory & Mechanics

### What is Graph RAG?

Graph RAG combines knowledge graphs with RAG to improve retrieval and generation.

**Architecture:**
1. **Query Processing**: Extract entities from query
2. **Graph Retrieval**: Find relevant subgraphs
3. **Vector Retrieval**: Find similar documents (optional)
4. **Hybrid Merge**: Combine graph + vector results
5. **Context Formatting**: Convert graph to text
6. **LLM Generation**: Generate response with graph context

**Graph Retrieval Process:**
```
Query: "What does Alice work on?"
    ↓
Extract entities: [Alice]
    ↓
Find nodes: (Person:Alice)
    ↓
Traverse relationships: (Alice)-[WORKS_ON]->(Project:X)
    ↓
Retrieve subgraph: Alice, Project X, related entities
    ↓
Format for LLM: "Alice works on Project X, which involves..."
```

### Hybrid Retrieval

**Vector Retrieval:**
- Embed query and documents
- Find top-K similar documents
- Good for semantic similarity

**Graph Retrieval:**
- Find entities in query
- Traverse graph from entities
- Good for relationship queries

**Hybrid Approach:**
- Run both retrievals
- Merge and rank results
- Deduplicate
- Best of both worlds

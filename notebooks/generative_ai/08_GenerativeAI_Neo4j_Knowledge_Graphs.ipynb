## What are Knowledge Graphs?

**Knowledge Graphs** represent information as networks of entities (nodes) and their relationships (edges). Unlike traditional databases, knowledge graphs emphasize relationships.

### Key Concepts

**Nodes (Vertices)**: Entities in the graph
- Examples: Person, Company, Document, Concept
- Have properties (attributes): name, age, type, etc.
- Can have labels (types): Person, Organization, etc.

**Relationships (Edges)**: Connections between nodes
- Examples: KNOWS, WORKS_FOR, CONTAINS, SIMILAR_TO
- Have direction: A → KNOWS → B (A knows B)
- Can have properties: since, weight, etc.

**Property Graph Model**: Nodes and relationships with properties
- Nodes: (Person {name: "Alice", age: 30})
- Relationships: (Alice)-[KNOWS {since: 2020}]->(Bob)
- Both can have multiple properties

**Cypher Query Language**: SQL-like language for graphs
- MATCH: Find patterns in graph
- CREATE: Add nodes/relationships
- WHERE: Filter results
- RETURN: Return results

### Why Use Knowledge Graphs?

1. **Relationship-Focused**: Natural for modeling connections
2. **Flexible Schema**: Easy to add new node/relationship types
3. **Traversal**: Fast path finding (e.g., shortest path)
4. **Semantic Understanding**: Explicit relationships between concepts
5. **Complement LLMs**: Provides structured knowledge to LLMs

### When to Use Knowledge Graphs

✅ **Good for:**
- Modeling relationships (social networks, dependencies)
- Complex queries involving paths and connections
- When relationships are as important as entities
- Knowledge bases with rich interconnections
- Recommendation systems (find related items)
- Fraud detection (find suspicious patterns)
- Complementing LLMs with structured knowledge

❌ **Not ideal for:**
- Simple tabular data (use SQL database)
- When relationships aren't important
- Very simple queries (overhead not worth it)
- When you need ACID transactions at scale
- When data is mostly unstructured text


## Theory & Mechanics

### What are Knowledge Graphs?

Knowledge graphs represent information as networks of nodes (entities) and edges (relationships).

**Example:**
```
(Person:Alice {age: 30})
    -[KNOWS {since: 2020}]-> (Person:Bob {age: 28})
    -[WORKS_FOR]-> (Company:TechCorp {founded: 2010})
```

**Key Operations:**
1. **Create**: Add nodes and relationships
2. **Match**: Find patterns in graph
3. **Traverse**: Follow relationships
4. **Query**: Extract information using patterns

### Neo4j Architecture

**Property Graph Model:**
- Nodes: Entities with labels and properties
- Relationships: Directed connections with types and properties
- Both can have multiple properties (key-value pairs)

**Cypher Query Language:**
- Declarative: Describe what you want, not how
- Pattern matching: Find patterns in graph
- Similar to SQL but for graphs

**Example Cypher:**
```cypher
MATCH (p:Person)-[r:KNOWS]->(friend:Person)
WHERE p.name = "Alice"
RETURN friend.name, r.since
```

### Graph vs. Relational Database

**Relational (SQL):**
- Tables with rows and columns
- Joins for relationships
- Fixed schema
- Good for structured data

**Graph (Neo4j):**
- Nodes and relationships
- Direct connections (no joins needed)
- Flexible schema
- Good for relationship-heavy data

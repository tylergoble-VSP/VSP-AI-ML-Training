## What is Chain-of-Thought?

**Chain-of-Thought (CoT)** is a prompting technique that encourages LLMs to show their reasoning process step-by-step before giving the final answer. Instead of jumping directly to the answer, the model explains its thinking.

### Key Concepts

**Standard Prompting**: Direct question → Direct answer
- Question: "What is 15 × 23?"
- Answer: "345"
- Problem: Model may make calculation errors without showing work

**Chain-of-Thought Prompting**: Question → Reasoning steps → Answer
- Question: "What is 15 × 23? Let's think step by step."
- Reasoning: "15 × 23 = 15 × (20 + 3) = 15 × 20 + 15 × 3 = 300 + 45 = 345"
- Answer: "345"
- Benefit: Breaking down problem helps model reason correctly

**Self-Consistency**: Generate multiple reasoning paths and take majority vote
- Generate several CoT solutions
- Extract final answers from each
- Take most common answer (consensus)
- Improves accuracy by reducing errors

### Why CoT Works

1. **Decomposition**: Breaks complex problems into simpler steps
2. **Error Detection**: Can spot mistakes in intermediate steps
3. **Verification**: Can check each step before proceeding
4. **Interpretability**: Shows how model arrived at answer
5. **Few-Shot Learning**: Examples teach model the reasoning pattern

### When to Use Chain-of-Thought

✅ **Good for:**
- Mathematical problems (arithmetic, algebra, word problems)
- Logical reasoning (puzzles, riddles, logic problems)
- Multi-step problems (planning, analysis)
- When you need to understand model's reasoning
- Complex problem-solving tasks
- When standard prompting gives incorrect answers

❌ **Not ideal for:**
- Simple factual questions (overhead not worth it)
- When you need fast, direct answers
- Very simple tasks (CoT adds unnecessary complexity)
- When model is already performing well


## Theory & Mechanics

### What is Chain-of-Thought?

CoT prompting encourages step-by-step reasoning by providing examples or instructions that show the model how to break down problems.

**Two Approaches:**

1. **Zero-Shot CoT**: Add "Let's think step by step" to prompt
   - Simple: Just add instruction
   - Works on many models
   - Less reliable than few-shot

2. **Few-Shot CoT**: Provide examples of step-by-step reasoning
   - More reliable
   - Teaches model the pattern
   - Requires example problems

**Example (Few-Shot CoT):**

```
Q: Roger has 5 tennis balls. He buys 2 more cans of tennis balls. Each can has 3 balls. How many does he have now?
A: Roger started with 5 balls. He bought 2 cans × 3 balls = 6 balls. Total: 5 + 6 = 11 balls.

Q: The cafeteria had 23 apples. They used 20 for lunch and bought 6 more. How many do they have?
A: They started with 23. Used 20, so 23 - 20 = 3 left. Bought 6 more: 3 + 6 = 9 apples.

Q: [Your question]
A: [Model will follow the pattern]
```

### Self-Consistency

**Self-Consistency** improves CoT by:
1. Generating multiple reasoning paths (different CoT solutions)
2. Extracting final answers from each path
3. Taking majority vote (most common answer)
4. Reduces impact of reasoning errors

**Why it works:**
- Different reasoning paths may have different errors
- Majority vote filters out errors
- More reliable than single CoT solution

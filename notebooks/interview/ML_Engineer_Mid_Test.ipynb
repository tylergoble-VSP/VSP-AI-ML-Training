- **`confusion_matrix`**: Shows prediction counts: true positives, false positives, true negatives, false negatives.
  - Rows = actual classes, Columns = predicted classes

### Classifier Algorithms

All classifiers in sklearn follow the same pattern:
1. **Instantiate**: `model = SomeClassifier(parameters)`
2. **Train**: `model.fit(X_train, y_train)`
3. **Predict**: `predictions = model.predict(X_test)`

**Available classifiers:**

- **`LogisticRegression`**: Despite the name, this is a classification algorithm.
  - Uses a logistic (sigmoid) function to predict probabilities
  - Good baseline, works well when classes are linearly separable
  - Benefits from scaled features

- **`DecisionTreeClassifier`**: Makes decisions by splitting data based on feature thresholds.
  - Easy to interpret (you can visualize the tree)
  - Does NOT require feature scaling
  - Can overfit if tree grows too deep

- **`RandomForestClassifier`**: An ensemble of many decision trees.
  - Each tree is trained on a random subset of data and features
  - Final prediction = majority vote of all trees
  - More robust than a single tree, reduces overfitting
  - Does NOT require feature scaling

- **`KNeighborsClassifier`**: Classifies based on the K nearest neighbors in feature space.
  - Finds the K training samples closest to the new point
  - Predicts the most common class among those K neighbors
  - REQUIRES feature scaling (uses distance calculations)

- **`SVC`**: Support Vector Classifier (Support Vector Machine for classification).
  - Finds the hyperplane that best separates classes
  - Can use kernel tricks for non-linear boundaries
  - Benefits significantly from feature scaling

## Question 1.1: List Comprehension with Filtering

### Background: What is a List Comprehension?

A **list comprehension** is a concise way to create lists in Python. It combines a loop and an optional condition into a single line.

**Basic syntax:**
```python
[expression for item in iterable]
```

**With filtering (conditional):**
```python
[expression for item in iterable if condition]
```

**How it works:**
1. `for item in iterable` - Loop through each element
2. `if condition` - Only include items where this is True (optional)
3. `expression` - Transform the item (this becomes the new list element)

**Example:**
```python
# Traditional loop approach:
result = []
for x in [1, 2, 3, 4, 5]:
    if x > 2:
        result.append(x * 10)
# result = [30, 40, 50]

# Same thing as a list comprehension:
result = [x * 10 for x in [1, 2, 3, 4, 5] if x > 2]
# result = [30, 40, 50]
```

### Background: Working with Dictionaries

A **dictionary** (`dict`) is Python's built-in mapping type that stores key-value pairs.


### The Problem: Grouping and Aggregating

You have a **list of dictionaries** where each dictionary represents one experiment run. Multiple runs exist for each model type. Your task is to compute the **average accuracy per model**.

This is a common pattern called **group-by aggregation**:
1. **Group**: Organize items by a key (model name)
2. **Aggregate**: Compute a summary statistic for each group (average)

### Approaches You Might Use

**Approach 1: Manual loop with dictionary**
```python
# Step 1: Collect all values per group
groups = {}
for item in data:
    key = item["model"]
    if key not in groups:
        groups[key] = []
    groups[key].append(item["accuracy"])

# Step 2: Compute average for each group
averages = {}
for key, values in groups.items():
    averages[key] = sum(values) / len(values)
```

**Approach 2: Using collections.defaultdict**
```python
from collections import defaultdict
groups = defaultdict(list)  # Auto-creates empty list for new keys
for item in data:
    groups[item["model"]].append(item["accuracy"])
```

**Approach 3: Using pandas**
```python
df = pd.DataFrame(data)
result = df.groupby("model")["accuracy"].mean().to_dict()
```

### Things to Consider

1. **Average formula**: `sum(values) / len(values)` or `sum(values) / count`
2. **Handling new keys**: Need to initialize storage for first occurrence of each model
3. **Code clarity vs brevity**: All approaches are valid; choose what you're comfortable with

### Why This Matters in ML

- Aggregating metrics across multiple experiment runs
- Computing mean and standard deviation of cross-validation scores
- Summarizing hyperparameter search results

## Question 1.3: Simple Class with Validation

### Background: Object-Oriented Programming in Python

A **class** is a blueprint for creating objects. Objects combine data (attributes) and behavior (methods).

**Key concepts:**

1. **`__init__` method (Constructor)**: Called automatically when you create an instance.
   ```python
   class Dog:
       def __init__(self, name, age):
           self.name = name  # Instance attribute
           self.age = age
   
   my_dog = Dog("Buddy", 3)  # __init__ is called here
   print(my_dog.name)  # "Buddy"
   ```

2. **`self` parameter**: Refers to the current instance. Must be the first parameter of instance methods.
   - `self.attribute` accesses/sets instance data
   - `self.method()` calls another method on the same instance

3. **Instance methods**: Functions defined inside a class that operate on instance data.
   ```python
   def is_adult(self):
       return self.age >= 2
   ```

4. **`__str__` method (Dunder/Magic method)**: Defines the string representation of an object.
   - Called by `print(obj)` and `str(obj)`
   - Should `return` a string (not print it)
   ```python
   def __str__(self):
       return f"Dog named {self.name}, age {self.age}"
   ```

5. **Raising exceptions**: Use `raise` to signal errors.
   ```python
   if age < 0:
       raise ValueError("Age cannot be negative")
   ```

6. **Default parameter values**: Provide fallback values for optional arguments.
   ```python
   def greet(self, greeting="Hello"):
       return f"{greeting}, {self.name}!"
   
   dog.greet()           # Uses default: "Hello, Buddy!"
   dog.greet("Hi there") # Uses provided: "Hi there, Buddy!"
   ```

# Part 2: ML Pipeline Implementation (~12 minutes)

In this section, you'll build a classification pipeline from start to finish.

### What is Classification?

**Classification** is a type of supervised learning where the goal is to predict a **categorical label** (discrete class) for each input.

- **Input**: Feature vector (measurements/attributes of a sample)
- **Output**: Class label (which category the sample belongs to)

**Examples:**
- Email → Spam or Not Spam (binary classification)
- Image → Cat, Dog, or Bird (multi-class classification)
- Flower measurements → Species (what we'll do here)

### The Iris Dataset

The Iris dataset is one of the most famous datasets in machine learning, introduced by statistician Ronald Fisher in 1936.

**Dataset details:**
- **150 samples** total (50 per species)
- **4 features** (measurements in centimeters):
  1. Sepal length
  2. Sepal width
  3. Petal length
  4. Petal width
- **3 classes** (species):
  - 0: Iris setosa
  - 1: Iris versicolor
  - 2: Iris virginica

**Why it's good for learning:**
- Small enough to train quickly
- Clean data (no missing values)
- Multi-class problem (more interesting than binary)
- Well-understood, easy to verify your results

### The ML Pipeline You'll Build

```
Raw Data → Split → (Optional: Scale) → Train Model → Evaluate → Compare
```

1. **Load data**: Get features (X) and labels (y)
2. **Split data**: Separate into training and test sets
3. **Scale features** (optional): Normalize feature ranges
4. **Train model**: Fit classifier on training data
5. **Evaluate**: Test on held-out data, compute metrics

#### Step 4: Feature Scaling (Optional but Important)

**What is scaling?**
Transforming features to have similar ranges/distributions.

**StandardScaler specifically:**
- Transforms each feature to have mean=0 and standard deviation=1
- Formula: `z = (x - mean) / std`
- Also called "standardization" or "z-score normalization"

**Why scale?**
Some algorithms are sensitive to feature magnitudes:
- **KNN**: Uses distance; larger-scale features dominate
- **SVM**: Optimization works better with similar scales
- **Logistic Regression**: Gradient descent converges faster
- **Neural Networks**: Similar reasons to logistic regression

**Which algorithms DON'T need scaling?**
- **Decision Trees**: Split on individual features, scale doesn't matter
- **Random Forest**: Ensemble of trees, same reasoning
- **Naive Bayes**: Uses probabilities, not distances

**CRITICAL: Fit on train, transform both**
```python
scaler = StandardScaler()

X_test_scaled = scaler.transform(X_test)        # Apply same transformation
```

**Why not fit on test data?**
- Test data represents "future unseen data"
- We wouldn't have access to its statistics in production
- Fitting on test = data leakage = artificially inflated performance

## Question 2.2: Model Training and Evaluation

### The Scikit-learn API Pattern

All scikit-learn models follow the same pattern:

```python

model = SomeClassifier(param1=value1, param2=value2)

# 2. Train the model on training data
model.fit(X_train, y_train)

# 3. Make predictions on new data
predictions = model.predict(X_test)

# 4. Evaluate predictions against true labels
```

### Understanding `accuracy_score`

**Definition:**
```
accuracy = (number of correct predictions) / (total number of predictions)
```

**Example:**
- 30 test samples
- 27 predicted correctly
- Accuracy = 27/30 = 0.90 (90%)

**Usage:**
```python
from sklearn.metrics import accuracy_score
accuracy = accuracy_score(y_true, y_predicted)
# y_true: The actual correct labels
# y_predicted: What your model predicted
```

| Classifier | Strengths | Considerations |

| **RandomForestClassifier** | Robust, handles non-linear, reduces overfitting | Slower than single tree; `n_estimators` controls number of trees |
| **KNeighborsClassifier** | Simple, no training phase | Slow prediction on large data; NEEDS scaling |
| **SVC** | Effective in high dimensions | Slower on large datasets; NEEDS scaling |

## Question 2.3: Best Model Analysis

### Understanding the Confusion Matrix

A **confusion matrix** shows the counts of predictions vs actual labels.

**Structure (for 3-class problem):**
```
                    Predicted
                 Class0  Class1  Class2
Actual Class0      10      0       0      ← 10 Class0 samples, all predicted correctly
       Class1       0      9       1      ← 10 Class1 samples, 9 correct, 1 confused with Class2  
       Class2       0      2       8      ← 10 Class2 samples, 8 correct, 2 confused with Class1
```

**Reading the matrix:**
- **Diagonal (top-left to bottom-right)**: Correct predictions
- **Off-diagonal**: Errors (misclassifications)
- **Rows**: Actual classes
- **Columns**: Predicted classes

**Usage:**
```python
from sklearn.metrics import confusion_matrix
cm = confusion_matrix(y_test, predictions)
print(cm)
```

### Understanding the Classification Report

The **classification report** provides per-class metrics:

**Metrics explained:**

1. **Precision**: Of all samples I predicted as class X, what fraction actually were class X?
   - Formula: `True Positives / (True Positives + False Positives)`
   - High precision = few false alarms
   - Example: Precision 0.90 for "spam" means 90% of emails I labeled spam were actually spam

2. **Recall** (also called Sensitivity): Of all actual class X samples, what fraction did I correctly identify?
   - Formula: `True Positives / (True Positives + False Negatives)`
   - High recall = few missed cases
   - Example: Recall 0.80 for "spam" means I caught 80% of all spam emails

3. **F1-Score**: Harmonic mean of precision and recall
   - Formula: `2 * (precision * recall) / (precision + recall)`
   - Balances precision and recall
   - Useful when you care about both equally

4. **Support**: Number of actual samples for that class in the test set

**Usage:**
```python
from sklearn.metrics import classification_report
print(classification_report(y_test, predictions, target_names=['setosa', 'versicolor', 'virginica']))
```

**Sample output:**
```

      setosa       1.00      1.00      1.00        10
  versicolor       0.90      0.90      0.90        10
```

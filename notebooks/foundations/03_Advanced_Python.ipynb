## 1. Iterators and Generators


```python
# Generator function using yield
def count_up_to(max_count):
    """Generator that counts up to max_count."""
    count = 1
    while count <= max_count:
        yield count  # Yield pauses execution and returns value
        count += 1

# Use the generator
for num in count_up_to(5):
    print(num)

# Generator expression (similar to list comprehension but lazy)
squares_gen = (x**2 for x in range(10))
print(f"Generator object: {squares_gen}")
print(f"First few values: {[next(squares_gen) for _ in range(5)]}")
```

## 2. Decorators


```python
from src.utils.timing import timeit

# Using the timeit decorator from our utilities
@timeit
def slow_function():
    """A function that takes some time."""
    total = sum(range(1000000))
    return total

result = slow_function()
print(f"Result: {result}")

# Custom decorator example
def uppercase_decorator(func):
    """Decorator that converts function output to uppercase."""
    def wrapper(*args, **kwargs):
        result = func(*args, **kwargs)
        return result.upper()
    return wrapper

@uppercase_decorator
def greet(name):
    return f"Hello, {name}!"

print(greet("Alice"))
```

```python
from src.utils.timing import TimeBlock

# Using context manager for file operations
with open('outputs/results/example.txt', 'w') as f:
    f.write("This file is automatically closed\n")

# Using our TimeBlock context manager
with TimeBlock("example_operation"):
    total = sum(range(100000))
    print(f"Sum: {total}")
```

## 1. Iterators and Generators

**What are Iterators?**
Iterators are objects that can be looped over. They produce values one at a time, which is memory-efficient for large datasets.

**What are Generators?**
Generators are a special type of iterator created using `yield` instead of `return`. They generate values on-the-fly (lazy evaluation) rather than storing them all in memory.

**Key Concepts:**
- **Iterator Protocol**: Objects with `__iter__()` and `__next__()` methods
- **Generator Functions**: Functions that use `yield` to produce values
- **Generator Expressions**: Like list comprehensions, but create generators (lazy)
- **Lazy Evaluation**: Values are computed only when needed

**Why use Generators?**
- **Memory Efficient**: Don't store all values in memory at once
- **Infinite Sequences**: Can represent sequences that never end
- **Performance**: Faster for large datasets
- **Composable**: Can chain generators together


```python
# ============================================
# GENERATOR FUNCTIONS: Using yield
# ============================================

# Generator functions use 'yield' instead of 'return'
# 'yield' pauses the function and returns a value, but remembers where it left off
# When called again, execution resumes from after the yield

def count_up_to(max_count):
    """
    Generator that counts up to max_count.
    
    This is a generator function because it uses 'yield'.
    Each time you ask for the next value, it continues from where it paused.
    
    Parameters:
        max_count: Maximum number to count to
    """
    count = 1  # Start counting from 1
    
    # Loop until we reach max_count
    while count <= max_count:
        # yield pauses execution here and returns 'count'
        # When next value is requested, execution resumes from the next line
        yield count  # Return count, but don't exit function - pause here
        
        # Execution resumes here when next value is requested
        count += 1  # Increment count for next iteration
    
    # When loop ends, generator is exhausted (no more values)

# ============================================
# USING GENERATORS: Iterating Over Values
# ============================================

# Generators can be used in for loops just like lists
# The for loop automatically calls next() to get each value
for num in count_up_to(5):
    # Each iteration gets the next value from the generator
    # First iteration: num = 1
    # Second iteration: num = 2
    # ... and so on
    print(num)  # Output: 1, 2, 3, 4, 5 (one per line)

# ============================================
# GENERATOR EXPRESSIONS: Lazy List Comprehensions
# ============================================

# Generator expressions are like list comprehensions but create generators
# Syntax: (expression for item in iterable) - note the parentheses, not brackets
# They're "lazy" - values are computed only when requested

# This creates a generator, not a list
# Values are NOT computed yet - they're computed on-demand
squares_gen = (x**2 for x in range(10))  # Generator expression
print(f"Generator object: {squares_gen}")  # Shows it's a generator object, not a list

# next() gets the next value from the generator
# Each call to next() computes the next value
# [next(squares_gen) for _ in range(5)] gets the first 5 values
print(f"First few values: {[next(squares_gen) for _ in range(5)]}")  # Output: [0, 1, 4, 9, 16]
```

## 2. Decorators

**What are Decorators?**
Decorators are functions that modify or enhance other functions without changing their code. They're a powerful way to add functionality (like logging, timing, caching) to functions.

**How Decorators Work:**
1. Decorator is a function that takes a function as input
2. Returns a new function (wrapper) that does something extra
3. The wrapper calls the original function and adds new behavior

**Syntax:** `@decorator_name` above a function definition

**Why use Decorators?**
- **Separation of Concerns**: Add functionality without modifying original code
- **Reusability**: Apply same enhancement to multiple functions
- **Clean Code**: Keep functions focused on their main purpose
- **Non-invasive**: Don't need to change function internals


```python
# ============================================
# USING DECORATORS: The @ Syntax
# ============================================

# Import the timeit decorator from our utilities module
# This decorator automatically measures how long a function takes to execute
from src.utils.timing import timeit

# The @timeit syntax applies the decorator to the function below
# @timeit is equivalent to: slow_function = timeit(slow_function)
@timeit  # This decorator wraps slow_function with timing code
def slow_function():
    """
    A function that takes some time.
    
    This function computes the sum of numbers from 0 to 999,999.
    The @timeit decorator automatically measures how long this takes.
    """
    # sum(range(1000000)) adds all numbers from 0 to 999,999
    # This is computationally expensive, so it takes measurable time
    total = sum(range(1000000))  # Sum: 0+1+2+...+999999 = 499,999,500,000
    return total

# When we call slow_function(), the decorator:
# 1. Records start time
# 2. Calls the original function
# 3. Records end time
# 4. Logs the duration
# 5. Returns the function's result
result = slow_function()  # Function executes, timing is logged automatically
print(f"Result: {result}")  # Output: Result: 499999500000

# ============================================
# CREATING CUSTOM DECORATORS
# ============================================

# Decorators are just functions that take a function and return a new function
def uppercase_decorator(func):
    """
    Decorator that converts function output to uppercase.
    
    This decorator wraps a function so its return value is always uppercase.
    
    Parameters:
        func: The function to decorate
    
    Returns:
        A wrapper function that calls func and uppercases the result
    """
    # Define the wrapper function that will replace the original
    def wrapper(*args, **kwargs):
        """
        Wrapper function that calls the original and modifies the result.
        
        *args and **kwargs allow the wrapper to accept any arguments
        and pass them to the original function.
        """
        # Call the original function with all its arguments
        result = func(*args, **kwargs)  # Execute the original function
        
        # Modify the result before returning
        return result.upper()  # Convert result to uppercase
    
    # Return the wrapper function
    # This wrapper will be used instead of the original function
    return wrapper

# Apply the decorator to a function
@uppercase_decorator  # This applies uppercase_decorator to greet()
def greet(name):
    """
    Greet someone by name.
    
    The @uppercase_decorator ensures the return value is always uppercase.
    """
    return f"Hello, {name}!"  # Returns "Hello, Alice!"

# When we call greet(), the decorator:
# 1. Calls the original greet() function
# 2. Gets the result: "Hello, Alice!"
# 3. Converts it to uppercase: "HELLO, ALICE!"
# 4. Returns the uppercase version
print(greet("Alice"))  # Output: HELLO, ALICE! (uppercase due to decorator)
```

## 3. Context Managers

**What are Context Managers?**
Context managers ensure proper setup and cleanup of resources (files, database connections, locks, etc.) using the `with` statement.

**How They Work:**
- `__enter__()`: Called when entering the `with` block (setup)
- `__exit__()`: Called when exiting the `with` block (cleanup, even if error occurs)

**Why use Context Managers?**
- **Automatic Cleanup**: Resources are always released, even if errors occur
- **Exception Safety**: Cleanup happens even when code crashes
- **Cleaner Code**: No need for try/finally blocks
- **Prevents Leaks**: Ensures files are closed, locks are released, etc.


```python
# ============================================
# CONTEXT MANAGERS: The 'with' Statement
# ============================================

# Import TimeBlock context manager from our utilities
from src.utils.timing import TimeBlock

# ============================================
# FILE CONTEXT MANAGER: Automatic File Closing
# ============================================

# The 'with' statement is a context manager
# open() returns a file object that acts as a context manager
# When you exit the 'with' block, the file is automatically closed

# 'w' mode means "write" - creates new file or overwrites existing
with open('outputs/results/example.txt', 'w') as f:
    # f is the file object
    # Inside this block, we can write to the file
    f.write("This file is automatically closed\n")  # Write text to file
    
# File is automatically closed here (when exiting the 'with' block)
# Even if an error occurred, the file would still be closed
# This prevents file handle leaks

# ============================================
# CUSTOM CONTEXT MANAGER: TimeBlock
# ============================================

# TimeBlock is a custom context manager that measures execution time
# It automatically logs how long code takes to execute

# The 'with' statement calls TimeBlock.__enter__() at the start
# and TimeBlock.__exit__() at the end
with TimeBlock("example_operation"):
    # This code block is timed
    # TimeBlock records start time when entering
    
    # Perform some computation
    total = sum(range(100000))  # Sum numbers from 0 to 99,999
    print(f"Sum: {total}")  # Output: Sum: 4999950000
    
    # TimeBlock records end time when exiting
    # Duration is automatically logged to a CSV file
```

## 4. Introduction to NumPy

**What is NumPy?**
NumPy (Numerical Python) is the fundamental library for numerical computing in Python. It provides efficient arrays and mathematical operations.

**Key Concepts:**
- **Arrays**: Multi-dimensional arrays (like lists, but faster and more powerful)
- **Vectorization**: Operations applied to entire arrays at once (no loops needed)
- **Broadcasting**: Automatic alignment of arrays with different shapes
- **Universal Functions (ufuncs)**: Fast element-wise operations

**Why NumPy?**
- **Speed**: 10-100x faster than Python lists for numerical operations
- **Memory Efficient**: Arrays use less memory than lists
- **Mathematical Operations**: Built-in functions for math, linear algebra, etc.
- **Foundation**: Used by pandas, scikit-learn, and most scientific Python libraries


```python
# ============================================
# NUMPY: Numerical Computing Library
# ============================================

# Import NumPy (convention: import as 'np' for brevity)
import numpy as np

# Import our normalization utility function
from src.processing.preprocessing import normalize_array

# ============================================
# CREATING NUMPY ARRAYS
# ============================================

# np.array() converts Python lists to NumPy arrays
# Arrays are like lists but optimized for numerical operations

# 1D array (one dimension): like a list, but with NumPy superpowers
arr1 = np.array([1, 2, 3, 4, 5])  # Creates array: [1 2 3 4 5]
print(f"1D Array: {arr1}")  # Display the array

# 2D array (two dimensions): like a matrix or table
# [[1, 2, 3], [4, 5, 6]] creates a 2x3 matrix (2 rows, 3 columns)
arr2 = np.array([[1, 2, 3], [4, 5, 6]])
print(f"2D Array:\n{arr2}")  # Display the 2D array

# .shape returns the dimensions of the array
# (2, 3) means 2 rows and 3 columns
print(f"Shape: {arr2.shape}")  # Output: Shape: (2, 3)

# ============================================
# VECTORIZED OPERATIONS: Fast Array Math
# ============================================

# NumPy operations are "vectorized" - they work on entire arrays at once
# This is MUCH faster than Python loops

# arr1 ** 2 squares every element in the array
# No loop needed! NumPy does it efficiently in C code
# Equivalent to: [x**2 for x in arr1] but much faster
squared = arr1 ** 2  # [1, 2, 3, 4, 5] ** 2 = [1, 4, 9, 16, 25]
print(f"Squared: {squared}")  # Output: Squared: [ 1  4  9 16 25]

# ============================================
# USING UTILITY FUNCTIONS: Normalization
# ============================================

# normalize_array() standardizes data (mean=0, std=1)
# This is important for many machine learning algorithms
normalized = normalize_array(arr1)  # Normalize arr1
print(f"Normalized: {normalized}")  # Output: Normalized array with mean≈0, std=1
```

## 5. Introduction to Pandas

**What is Pandas?**
Pandas is the primary library for data manipulation and analysis in Python. It provides DataFrames (tables) and Series (columns) for working with structured data.

**Key Concepts:**
- **DataFrame**: 2D table with rows and columns (like Excel spreadsheet)
- **Series**: 1D array with labels (like a single column)
- **Index**: Row labels (like row numbers, but can be custom)
- **Operations**: Filtering, grouping, merging, aggregating data

**Why Pandas?**
- **Data Analysis**: Perfect for working with CSV files, databases, etc.
- **Data Cleaning**: Easy to handle missing values, duplicates, etc.
- **Powerful Operations**: Group by, pivot, merge, aggregate
- **Integration**: Works seamlessly with NumPy, matplotlib, scikit-learn


```python
# ============================================
# PANDAS: Data Manipulation Library
# ============================================

# Import pandas (convention: import as 'pd' for brevity)
import pandas as pd

# ============================================
# CREATING A DATAFRAME: From Dictionary
# ============================================

# DataFrames can be created from dictionaries
# Keys become column names, values become column data
# All lists must have the same length (same number of rows)

data = {
    'name': ['Alice', 'Bob', 'Charlie'],      # Column 1: names
    'age': [25, 30, 35],                      # Column 2: ages
    'city': ['New York', 'London', 'Tokyo']   # Column 3: cities
}
# This creates a table with 3 rows and 3 columns

# pd.DataFrame() creates a DataFrame from the dictionary
df = pd.DataFrame(data)  # Convert dictionary to DataFrame

print("DataFrame:")
print(df)  # Display the entire DataFrame (table format)

# ============================================
# DATAFRAME METHODS: Exploring Data
# ============================================

# .describe() provides statistical summary of numerical columns
# Shows: count, mean, std, min, max, quartiles
print(f"\nBasic info:")
print(df.describe())  # Statistical summary (only for numeric columns like 'age')

# ============================================
# ACCESSING COLUMNS: Series Objects
# ============================================

# df['column_name'] accesses a single column
# Returns a Series (1D array with labels)
print(f"\nAge column:")
print(df['age'])  # Output: Series with ages [25, 30, 35] and row indices [0, 1, 2]
```

# Embedding Models

This module introduces embedding models for representing complex data (like text) in continuous vector spaces.

## Learning Objectives

- Understand what embeddings are
- Use pre-trained word embeddings
- Work with sentence transformers
- Visualize embeddings


```python
from src.llm.embeddings import cosine_similarity, average_word_embeddings
import numpy as np

# Example: Simple word embeddings (demonstration)
# In practice, you would use pre-trained models like Word2Vec, BERT, etc.

# Create simple example embeddings
word_embeddings = {
    'king': np.array([0.5, 0.3, 0.2]),
    'queen': np.array([0.4, 0.4, 0.2]),
    'man': np.array([0.6, 0.2, 0.2]),
    'woman': np.array([0.3, 0.5, 0.2])
}

# Calculate similarity
king_queen_sim = cosine_similarity(word_embeddings['king'], word_embeddings['queen'])
print(f"King-Queen similarity: {king_queen_sim:.3f}")

# Using sentence transformers (if available)
try:
    from sentence_transformers import SentenceTransformer
    model = SentenceTransformer('all-MiniLM-L6-v2')
    
    sentences = ["The cat sat on the mat", "A cat is sitting on a mat"]
    embeddings = model.encode(sentences)
    
    similarity = cosine_similarity(embeddings[0], embeddings[1])
    print(f"\nSentence similarity: {similarity:.3f}")
except ImportError:
    print("\nNote: Install sentence-transformers for full functionality")
    print("pip install sentence-transformers")
```

## What are Embeddings?

**Embeddings** are dense vector representations of discrete objects (words, sentences, images, etc.) in a continuous vector space. They capture semantic meaning and relationships.

### Key Concepts

**Word Embeddings**: Vector representations of words
- Each word is mapped to a fixed-size vector (e.g., 100, 300 dimensions)
- Similar words have similar vectors (close in vector space)
- Captures semantic relationships (king - man + woman ≈ queen)
- Learned from large text corpora

**Sentence Embeddings**: Vector representations of entire sentences
- Encodes meaning of full sentences, not just individual words
- Captures context and word order
- Useful for semantic similarity, search, clustering

**Cosine Similarity**: Measure of similarity between vectors
- cos(θ) = (A · B) / (||A|| × ||B||)
- Range: -1 to 1 (1 = identical, 0 = orthogonal, -1 = opposite)
- Measures angle between vectors, not magnitude

### Why Use Embeddings?

1. **Machine Learning Compatibility**: ML models need numerical inputs
2. **Semantic Understanding**: Captures meaning, not just spelling
3. **Dimensionality**: Compact representation (vs. one-hot encoding)
4. **Transfer Learning**: Pre-trained embeddings work well for many tasks
5. **Similarity**: Easy to find similar words/sentences

### When to Use Embeddings

✅ **Good for:**
- Natural language processing tasks
- Text classification and clustering
- Semantic search and recommendation
- Transfer learning for NLP
- When you need to capture semantic meaning
- Working with categorical data that has relationships

❌ **Not ideal for:**
- Very small vocabularies (overhead not worth it)
- When exact matches are required (not semantic similarity)
- Real-time applications with strict latency requirements
- When interpretability is critical (embeddings are black boxes)


```python
# ============================================
# IMPORTING LIBRARIES: Setting Up Our Tools
# ============================================

# Our custom embedding utilities
from src.llm.embeddings import cosine_similarity, average_word_embeddings
# cosine_similarity(): Calculate cosine similarity between two vectors
# average_word_embeddings(): Average word embeddings to create sentence embedding

# NumPy: Numerical computing
import numpy as np  # NumPy: Arrays and mathematical operations

# ============================================
# WORD EMBEDDINGS: Understanding Vector Representations
# ============================================

# Example: Simple word embeddings (demonstration)
# In practice, you would use pre-trained models like Word2Vec, GloVe, BERT, etc.
# Pre-trained embeddings are learned from millions of documents
# They capture semantic relationships (similar words have similar vectors)

# Create simple example embeddings for demonstration
# Real embeddings have 100-300+ dimensions and capture complex relationships
word_embeddings = {
    'king': np.array([0.5, 0.3, 0.2]),    # Vector representation of "king"
    'queen': np.array([0.4, 0.4, 0.2]),   # Vector representation of "queen"
    'man': np.array([0.6, 0.2, 0.2]),     # Vector representation of "man"
    'woman': np.array([0.3, 0.5, 0.2])   # Vector representation of "woman"
}
# Each word is represented as a vector (array of numbers)
# Similar words should have similar vectors
# In real embeddings: king and queen are close (both royalty)
#                     man and woman are close (both people)
#                     king - man + woman ≈ queen (famous word analogy!)

# ============================================
# CALCULATING SIMILARITY: Cosine Similarity
# ============================================

# Calculate similarity between "king" and "queen"
king_queen_sim = cosine_similarity(word_embeddings['king'], word_embeddings['queen'])
# cosine_similarity(): Measures how similar two vectors are
# Formula: cos(θ) = (A · B) / (||A|| × ||B||)
# - A · B: Dot product of vectors
# - ||A||, ||B||: Magnitude (length) of vectors
# Returns: Value between -1 and 1
#   - 1.0: Identical vectors (same direction)
#   - 0.0: Orthogonal vectors (no relationship)
#   - -1.0: Opposite vectors (opposite meaning)

print(f"King-Queen similarity: {king_queen_sim:.3f}")
# Higher similarity = more similar words
# In real embeddings, king-queen similarity is high (both royalty)

# ============================================
# SENTENCE EMBEDDINGS: Using Sentence Transformers
# ============================================

# Sentence transformers: Models that create embeddings for entire sentences
# More powerful than averaging word embeddings (captures context and word order)
# Pre-trained on large datasets to understand sentence meaning

try:
    # Try to import sentence-transformers library
    from sentence_transformers import SentenceTransformer
    # SentenceTransformer: Pre-trained models for sentence embeddings
    # These models are trained to understand semantic meaning of sentences
    
    # Load pre-trained model
    model = SentenceTransformer('all-MiniLM-L6-v2')
    # 'all-MiniLM-L6-v2': Lightweight, fast model (good for learning)
    # Other options: 'all-mpnet-base-v2' (more accurate, slower)
    # Model is downloaded automatically on first use
    
    # Example sentences to compare
    sentences = ["The cat sat on the mat", "A cat is sitting on a mat"]
    # These sentences have similar meaning (different wording)
    # Sentence embeddings should capture this similarity
    
    # Encode sentences to embeddings
    embeddings = model.encode(sentences)
    # model.encode(): Convert sentences to vector embeddings
    # Input: List of strings (sentences)
    # Output: NumPy array of shape (n_sentences, embedding_dim)
    # Each row is the embedding vector for one sentence
    # embeddings[0]: Vector for first sentence
    # embeddings[1]: Vector for second sentence
    
    # Calculate similarity between sentence embeddings
    similarity = cosine_similarity(embeddings[0], embeddings[1])
    # Compare embeddings of the two sentences
    # High similarity = sentences have similar meaning
    
    print(f"\nSentence similarity: {similarity:.3f}")
    # Should be high (>0.8) since sentences have similar meaning
    # Different wording but same semantic content
    
except ImportError:
    # If sentence-transformers is not installed, show installation instructions
    print("\nNote: Install sentence-transformers for full functionality")
    print("pip install sentence-transformers")
    # This library provides state-of-the-art sentence embeddings
    # Install with: pip install sentence-transformers

# ============================================
# KEY TAKEAWAYS
# ============================================

# 1. Embeddings convert discrete objects (words, sentences) to continuous vectors
# 2. Similar objects have similar embeddings (close in vector space)
# 3. Cosine similarity measures how similar two embeddings are
# 4. Pre-trained embeddings capture semantic relationships from large datasets
# 5. Sentence embeddings are more powerful than word embeddings (capture context)
# 6. Embeddings enable ML models to work with text data
```

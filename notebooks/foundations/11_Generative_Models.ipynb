## What are Generative Models?

**Generative models** learn the underlying probability distribution of training data and can generate new samples from that distribution. Unlike discriminative models (which classify or predict), generative models create new data.

### Generative vs Discriminative Models

**Discriminative Models** (what we've seen so far):
- Learn P(y | x): Probability of label given input
- Examples: Logistic Regression, CNNs, SVMs
- Goal: Classify or predict
- Answer: "Is this a cat or dog?"

**Generative Models** (this notebook):
- Learn P(x): Probability distribution of data
- Can generate new samples: x ~ P(x)
- Examples: VAEs, GANs, Autoregressive models
- Goal: Create new data
- Answer: "Generate a new image of a cat"

### Key Types of Generative Models

1. **Variational Autoencoders (VAE)**
   - Encoder: Maps data to latent space (compressed representation)
   - Decoder: Maps latent space back to data space
   - Learns smooth, continuous latent space
   - Can interpolate between samples
   - More stable training than GANs

2. **Generative Adversarial Networks (GAN)**
   - Generator: Creates fake data from noise
   - Discriminator: Distinguishes real from fake
   - Adversarial training (two networks compete)
   - Can produce very realistic samples
   - Training can be unstable

3. **Autoregressive Models**
   - Generate data step-by-step (e.g., one pixel at a time)
   - Examples: PixelRNN, GPT (for text)
   - Slow generation but high quality

### When to Use Generative Models

✅ **Good for:**
- Data augmentation (create more training data)
- Image generation and synthesis
- Anomaly detection (low probability = anomaly)
- Data imputation (fill in missing values)
- Creative applications (art, music, text generation)
- Understanding data distributions

❌ **Not ideal for:**
- Standard classification tasks (use discriminative models)
- Small datasets (need lots of data to learn distribution)
- When you need exact control over output
- Real-time applications (generation can be slow)
- When interpretability is critical

# Generative Models

This module explores generative models that can create new data samples, including Variational Autoencoders (VAE) and Generative Adversarial Networks (GANs).

## Learning Objectives

- Understand generative vs discriminative models
- Implement a simple VAE
- Understand GAN architecture
- Generate new data samples


```python
import torch
import torch.nn as nn
from torchvision import datasets, transforms
from src.models.generative import SimpleVAE, vae_loss
import matplotlib.pyplot as plt

# Note: This is a simplified example. Full implementation would require training.
print("Generative Models: VAE and GAN")

# Example VAE architecture
input_dim = 784  # 28x28 for MNIST
latent_dim = 20
vae = SimpleVAE(input_dim=input_dim, latent_dim=latent_dim)

print(f"\nVAE Model:")
print(f"Input dimension: {input_dim}")
print(f"Latent dimension: {latent_dim}")

# Example: Generate from random latent vector
vae.eval()
with torch.no_grad():
    # Sample random latent vector
    z = torch.randn(1, latent_dim)
    # Decode to generate image
    generated = vae.decode(z)
    print(f"\nGenerated sample shape: {generated.shape}")

print("\nNote: For full functionality, train the VAE on MNIST dataset.")
print("This requires PyTorch and torchvision installed.")
```

## 2. Variables and Types

Python is dynamically typed - you don't need to declare variable types. Python automatically determines the type based on the value assigned.


```python
# Numbers: integers and floats
age = 25  # Integer
height = 5.9  # Float (decimal number)
print(f"Age: {age}, Type: {type(age)}")
print(f"Height: {height}, Type: {type(height)}")

# Strings: text data
name = "Alice"
greeting = 'Hello, World!'  # Single or double quotes work
print(f"Name: {name}, Type: {type(name)}")

# Booleans: True or False
is_student = True
has_graduated = False
print(f"Is student: {is_student}, Type: {type(is_student)}")
```

## 3. Data Structures

Python provides several built-in data structures for organizing data.


```python
# Lists: ordered, mutable collections
fruits = ["apple", "banana", "orange"]
numbers = [1, 2, 3, 4, 5]
mixed = [1, "hello", 3.14, True]  # Can contain different types

print(f"Fruits: {fruits}")
print(f"First fruit: {fruits[0]}")  # Indexing starts at 0
print(f"Last fruit: {fruits[-1]}")  # Negative index from end

# Lists are mutable - you can change them
fruits.append("grape")  # Add to end
print(f"After append: {fruits}")

# Slicing: get a subset
print(f"First two: {fruits[0:2]}")
print(f"All but last: {fruits[:-1]}")
```

```python
# Tuples: ordered, immutable collections
coordinates = (10, 20)
point = (3.5, 4.2, 1.8)  # Can have multiple elements

print(f"Coordinates: {coordinates}")
print(f"X coordinate: {coordinates[0]}")

# Tuples are immutable - cannot change after creation
# coordinates[0] = 15  # This would cause an error!
```

```python
# Dictionaries: key-value pairs
student = {
    "name": "Alice",
    "age": 20,
    "grade": "A"
}

print(f"Student: {student}")
print(f"Name: {student['name']}")
print(f"Age: {student.get('age')}")  # Alternative way to access

# Add or modify entries
student["major"] = "Computer Science"
student["age"] = 21  # Update existing
print(f"Updated student: {student}")
```

```python
# Sets: unordered collections of unique elements
unique_numbers = {1, 2, 3, 4, 5}
colors = {"red", "green", "blue"}

print(f"Unique numbers: {unique_numbers}")

# Sets automatically remove duplicates
numbers_with_duplicates = {1, 2, 2, 3, 3, 3}
print(f"After removing duplicates: {numbers_with_duplicates}")

# Set operations
set1 = {1, 2, 3, 4}
set2 = {3, 4, 5, 6}
print(f"Union: {set1 | set2}")
print(f"Intersection: {set1 & set2}")
print(f"Difference: {set1 - set2}")
```

## 4. Control Flow

Control flow allows your program to make decisions and repeat actions.


```python
# If/elif/else statements
score = 85

if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 70:
    grade = "C"
else:
    grade = "F"

print(f"Score: {score}, Grade: {grade}")
```

```python
# For loops: iterate over sequences
fruits = ["apple", "banana", "orange"]

for fruit in fruits:
    print(f"I like {fruit}")

# Loop with range
for i in range(5):
    print(f"Number: {i}")

# Loop with enumerate (get index and value)
for index, fruit in enumerate(fruits):
    print(f"{index}: {fruit}")
```

```python
# While loops: repeat while condition is true
count = 0
while count < 5:
    print(f"Count: {count}")
    count += 1  # Increment count
```

```python
# Define a simple function
def greet(name):
    """Return a greeting message."""
    return f"Hello, {name}!"

# Call the function
message = greet("Alice")
print(message)
```

```python
# Function with multiple parameters
def add_numbers(a, b):
    """Add two numbers and return the result."""
    result = a + b
    return result

first_number = 5
second_number = 3

sum_result = add_numbers(first_number, second_number)
print(f"{first_number} + {second_number} = {sum_result}")
```

```python
# Function with default parameter
def power(base, exponent=2):
    """Raise base to the power of exponent."""
    return base ** exponent

print(f"2^3 = {power(2, 3)}")
print(f"5^2 = {power(5)}")  # Uses default exponent=2
```

Create a list of numbers from 1 to 10, then:

### Exercise 2: FizzBuzz

Write a function that prints numbers from 1 to n. For multiples of 3, print "Fizz". For multiples of 5, print "Buzz". For multiples of both, print "FizzBuzz".


```python
# Your solution here
def fizzbuzz(n):
    """Print FizzBuzz for numbers 1 to n."""
    for i in range(1, n + 1):
        if i % 3 == 0 and i % 5 == 0:
            print("FizzBuzz")
        elif i % 3 == 0:
            print("Fizz")
        elif i % 5 == 0:
            print("Buzz")
        else:
            print(i)

# Test with numbers 1 to 15
fizzbuzz(15)
```
